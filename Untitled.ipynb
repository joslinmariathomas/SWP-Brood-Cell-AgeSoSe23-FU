```python
import numpy as np
def count_long_subarrays(A: list):
    B = [A[i]<A[i+1] for i in range(len(A)-1)]
    B = [int(x) for x in B]
    print(B)
    C = [1 if (B[i]==1 & B[i+1]==0) else 0 for i in range(len(B)-1)]
    print(C)
    
    return None
```
